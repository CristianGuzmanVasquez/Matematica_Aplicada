Considere la sucesión $a_{n} = 3n^{2} + 7$. Por medio de un código en *Python* despliega en pantalla:

1. Los primeros $5$ términos.
2. Los $5$ términos que vienen inmediatamente después del decimoquinto término.

```python
1
#Creación de sucesiones con listas
sucesion = []
for i in range(5):
    n = i + 1
    sucesion.append(3*n**2+7)
    print(f'El término {n} es {sucesion[i]}')
```

# Problema 2

Considere la sucesión $g_{n} = 5n^{3}$. Implemente un código en *Python* que muestre en pantalla:

1. Los primeros $4$ términos de la sucesión.
2. Los $4$ términos que vienen inmediatamente después del octavo término.
3. Determinar si el término $40.000$ pertenece a la sucesión. Si es así, indicar en qué posición se encuentra.

```python
1
sucesion = []
for n in range (1,5):
    sucesion.append(5*n**3)

print('1) Los 4 términos son:')
print(sucesion)
```

```python
sucesion = []
for n in range (9,13):
    sucesion.append(5*n**3)

#print('2) Los  términos que vienen inmediatamente después del octavo término son:')
print(sucesion)
```

```python
3
#Determinar si un determinado valor es término de la sucesión. (CODIGO)
final=40000
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino= 5*n**3
    dif=final-termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```

# Problema 4

Implementa un código que permita guardar en una lista la sucesión de los primeros $n$ números pares, con $n \geq 1$. Luego, determina:

1. ¿Cuál es el valor del décimo término?
2. Calcula, utilizando un ciclo *for*, la suma de los primeros $100$ números pares.
3. ¿En qué posición se encuentra el número $58$ dentro de la sucesión?

```python
1
sucesion = []  

for n in range(1, 11):  
    sucesion.append(2*n)  

print(sucesion)  
```

El décimo término de la sucesión es 20.

```python
2
#Suma de términos de una lista      CODIGO PARA SUMAR
suma = 0  # Inicializa la suma en cero

for n in range(1, 101):  # Recorre los primeros 100 términos
    suma = suma + 2*n  # Va sumando cada número par acumular el nuevo término \(2n\) sobre lo que ya llevamos sumado.

print(f'La suma de los primeros 100 números pares es {suma}')
```

```python
#Determinar si un determinado valor es término de la sucesión. (CODIGO)
final=58
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=2*n
    dif=final-termino

if dif==0:    # COMPARA 
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
