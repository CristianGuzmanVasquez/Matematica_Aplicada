## A
Un nuevo software tiene $500$ usuarios y se espera que el número de usuarios crezca un $15\%$ respecto al mes anterior. Considere que $b_{n}$ es el número de usuarios transcurridos $n$ meses.

1. Calcula cuántos usuarios tendrá el software al cabo de tres meses. ¿Y al cabo de seis meses?
2. Mediante un código en *Pyhton* muestra la cantidad de usuarios que tendrá el software cada mes durante el primer año.
2. ¿Cuántos usuarios tendrá en total el software luego de un año? Utilice ciclos *for* para calcular.

```python
#-1

b_3 = 500*1.15**3
b_6 = 500*1.15**6
print(f"Al cabo de tres meses el software tendra {b_3:.0f} usuarios y al cabo de seis meses tendra {b_6:.0f} usuarios")
```

```python
#-2

sucesion = []
for n in range(1,13):
    sucesion.append(round(500*1.15**n))
    print(sucesion)
```

```python
#-3
suma = 0  # Inicializa la suma en cero

for n in range(1, 13):  # Recorre los primeros 100 términos
    suma = suma + 500*1.15**n  # Va sumando cada número par acumular el nuevo término \(2n\) sobre lo que ya llevamos sumado.

print(f'La cantidad total de usuarios en un año es de {suma:.0f}')
```


## Pregunta 2

Martina decidió que dejará de aumentar sus depósitos cuando el monto correspondiente a una semana alcance los **$100.000**.

**¿En qué semana su depósito alcanzará por primera vez los $100.000?**

Utiliza Python para determinar la posición correspondiente dentro de la sucesión e interrpete el resultado

```python
final=100000
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=5000*n+10000
    dif=final-termino

if dif==0:
    print(f'Martina alcanzara los ${final} en la semana {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
