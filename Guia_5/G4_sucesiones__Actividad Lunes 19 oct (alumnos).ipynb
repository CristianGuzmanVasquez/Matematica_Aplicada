## Modelo común para los ejercicios

Una empresa registra semanalmente el consumo de energía de una máquina. El consumo de la semana $n$, medido en kWh, se representa mediante:

$$a_n=48+0.7n,\qquad n=1,2,3,\ldots$$

Aquí, $n$ indica el número de semana. **La primera semana corresponde a $n=1$.**

## Ejercicio 1. Consumo en una semana determinada

La empresa necesita estimar el consumo de energía de la máquina durante la **semana 12**.

**Pregunta:** ¿Cuál será el consumo de energía, en kWh, durante la semana 12? 

```python
a_n = 48+0.7*12
print(f"a la semana 12 el consumo de energia sera de {a_n:.1f} KWh")
```

## Ejercicio 2. ¿Pertenece a la sucesión?

El encargado de operaciones desea saber si el consumo semanal de la máquina puede ser exactamente **55 kWh** según el modelo.

**Pregunta:** ¿El valor 55 pertenece a la sucesión? Si pertenece, indique en qué semana se alcanza. 

```python
final=55
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=48+0.7*n
    dif=final-termino

if dif==0:
    print(f'El consumo semanal de la maquina llegara a los {final} kWh en la semana {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```

## Ejercicio 3. Consumo total acumulado

Para planificar los gastos, la empresa necesita calcular el consumo **total de las primeras 10 semanas**.

**Pregunta:** ¿Cuántos kWh se consumen en total durante las primeras  10 semanas ?

```python
suma = 0    # Inicializa la suma en cero

for n in range(1 , 10 + 1):  # Recorre los primeros n términos
    suma = suma + 48+0.7*n   # Va sumando cada número 

print(f'El costo total de las primeras 10 semanas es de {suma:.1f} kWh')
```

## Ejercicio 4. Mostrar términos mediante `for`

El equipo de mantenimiento quiere visualizar el consumo de energía de **cada una de las primeras ocho semanas**.

**Pregunta:** Escriba un programa que utilice un ciclo `for` para mostrar en pantalla los consumos de las primeras 8 semanas.



```python
sucesion = []  # Crea una lista vacía para guardar los términos de la sucesión

for n in range(1, 9):  # Recorre los valores de n desde inicio hasta fin
    sucesion.append(round(48+0.7*n ))  # Calcula cada término y lo agrega a la lista

print(sucesion)
   
```

## Ejercicio 5. Encontrar la posición de un término

Considere la sucesión definida por:

$$a_n=n^3-2n$$

**Pregunta:** Determine el valor de $n$ para el cual $a_n=711$.



```python
final=711
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=n**3-2*n
    dif=final-termino

if dif==0:
    print(f'Para conseguir un valor de {final}, el valor de n debes ser de {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
